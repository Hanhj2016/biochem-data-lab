# 02 Control vs Treatment t-test - Python

## Biochemistry question

In this synthetic enzyme assay, is the mean enzyme activity different between the control and treatment groups?


```python
import pandas as pd
from scipy import stats

df = pd.read_csv('../data/enzyme_activity_sample.csv')
control = df[df['group'] == 'control']['enzyme_activity']
treatment = df[df['group'] == 'treatment']['enzyme_activity']

t_stat, p_value = stats.ttest_ind(control, treatment, equal_var=False)
print(f't statistic: {t_stat:.3f}')
print(f'p-value: {p_value:.4f}')
```

## Interpretation Questions

- What two groups does the t-test compare?
- Does the p-value alone prove biological importance?
- What limitation should be mentioned when reporting this result?

## Limitations

- This is a small synthetic dataset for learning.
- A t-test result is not evidence of clinical, diagnostic, or drug efficacy.
- Real experimental conclusions would need study design details, replication, and data quality checks.
