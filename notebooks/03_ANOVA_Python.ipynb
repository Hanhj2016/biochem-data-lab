# 03 ANOVA for Multiple Groups - Python

## Biochemistry question

In this synthetic cell viability dataset, do any of the group means appear different from the others?


```python
import plotly.io as pio
pio.renderers.default = "iframe"
```

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols
import plotly.express as px

df = pd.read_csv('../data/anova_cell_viability_sample.csv')
df.head()
```

```python
model = ols('cell_viability ~ C(group)', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
anova_table
```

```python
dose_order = ['control', 'low_dose', 'medium_dose', 'high_dose']
summary = df.groupby('group').agg(mean_viability=('cell_viability','mean'), sd_viability=('cell_viability','std'), n=('cell_viability','count')).reindex(dose_order).reset_index()
summary['sem_viability'] = summary['sd_viability'] / (summary['n'] ** 0.5)
fig = px.bar(summary, x='group', y='mean_viability', error_y='sem_viability', title='Cell Viability Across Dose Groups', category_orders={'group': dose_order})
# If this chart does not render in Jupyter, try: fig.show(renderer="browser")
fig.show(renderer="iframe")
```
