# 20 Sequence Basics: GC Content and Motifs - Python

## Biochemistry question

How can simple code summarize short synthetic DNA sequences?

This notebook uses synthetic sequences only. They are not real genomic, patient, clinical, or proprietary sequences.


```python
from pathlib import Path
import sys
import pandas as pd
import plotly.express as px

PROJECT_ROOT = Path("..").resolve()
sys.path.append(str(PROJECT_ROOT))
from src.sequence_basics import count_motif, reverse_complement, summarize_sequences

df = pd.read_csv(PROJECT_ROOT / "data" / "sequences" / "synthetic_sequences.csv")
df
```

## 1. Length and GC Content


```python
summary = summarize_sequences(df)
summary[["sequence_id", "description", "sequence_length", "gc_content_percent"]].round(1)
```

```python
fig = px.bar(summary, x="sequence_id", y="gc_content_percent", title="Synthetic Sequence GC Content", labels={"gc_content_percent":"GC content (%)"})
fig.show(renderer="iframe")
```

## 2. Motif Counts and Reverse Complements


```python
summary["ATG_count"] = summary["dna_sequence"].map(lambda seq: count_motif(seq, "ATG"))
summary["reverse_complement"] = summary["dna_sequence"].map(reverse_complement)
summary[["sequence_id", "ATG_count", "reverse_complement"]]
```

## What You Should Notice

- GC content differs across synthetic sequences.
- Motif searches depend on the exact sequence pattern selected.
- Reverse complements are useful for thinking about DNA strand orientation.
