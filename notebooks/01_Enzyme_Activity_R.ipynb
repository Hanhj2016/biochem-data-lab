# 01 Enzyme Activity Summary - R

## Biochemistry question

In this synthetic enzyme assay, how does enzyme activity compare between the control group and the treatment group?


```r
library(tidyverse)

df <- read_csv('../data/enzyme_activity_sample.csv')
df
```

```r
summary <- df %>%
  group_by(group) %>%
  summarise(mean_activity = mean(enzyme_activity),
            sd_activity = sd(enzyme_activity),
            n = n(),
            sem_activity = sd_activity / sqrt(n),
            .groups = 'drop')
summary
```

```r
ggplot(summary, aes(x = group, y = mean_activity)) +
  geom_col() +
  geom_errorbar(aes(ymin = mean_activity - sem_activity,
                    ymax = mean_activity + sem_activity), width = 0.2) +
  labs(title = 'Enzyme Activity by Group', x = 'Group', y = 'Mean Enzyme Activity')
```

## What you should notice

- The treatment group has a lower mean enzyme activity than the control group in this synthetic assay.
- The R workflow uses grouped summaries before plotting the means.

## Interpretation practice

- Which group has lower mean enzyme activity?
- What do the error bars help you compare?
- Which part of the R syntax feels clearest compared with Python?

## Common mistake

- Do not treat a smaller mean as proof of a mechanism. It is a pattern to describe cautiously.

## Limitations

- This is synthetic learning data, not a real experiment.
- More replicates and experimental context would be needed before making biological claims.
