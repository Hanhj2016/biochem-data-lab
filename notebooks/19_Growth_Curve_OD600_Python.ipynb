## Biochemistry question

How do synthetic OD600 measurements change over time for two microbial growth curves?

This notebook uses synthetic data for learning. It is not real culture or clinical microbiology data.


```python
import plotly.io as pio
pio.renderers.default = "iframe"
```

```python
from pathlib import Path
import sys
import pandas as pd
import plotly.express as px

PROJECT_ROOT = Path("..").resolve()
sys.path.append(str(PROJECT_ROOT))
from src.growth_curve import estimate_log_phase_rate, summarize_growth

df = pd.read_csv(PROJECT_ROOT / "data" / "growth_curve" / "bacterial_growth_curve.csv")
df.head()
```

## 1. Summarize Replicates


```python
summary = summarize_growth(df)
summary.head()
```

## 2. Plot OD600 Over Time


```python
fig = px.line(summary, x="time_hr", y="mean_od600", color="strain", markers=True, error_y="sem_od600", title="Synthetic OD600 Growth Curve")
fig.update_yaxes(title="Mean OD600")
# If this chart does not render in Jupyter, try: fig.show(renderer="browser")
fig.show(renderer="iframe")
```

## 3. Estimate Log-phase Growth Rate


```python
rates = estimate_log_phase_rate(summary, start_hr=2, end_hr=6)
rates.round(3)
```

## Industry / Lab Reality

In real growth-curve work, OD600 analysis needs blank correction, instrument range checks, culture conditions, inoculation details, and a clearly reported growth-rate window. The selected time window can change the estimated rate.

## Career / Job Skill Connection

This notebook practices skills used in microbiology, fermentation, bioprocessing, and synthetic biology projects: time-course summaries, growth-rate estimation, doubling-time interpretation, and metadata-aware reporting.


## What You Should Notice

- OD600 increases over time and begins to level off.
- Strains can differ in apparent growth rate and final density.
- The log-phase rate depends on the time window selected.

## Interpretation Practice

- Which strain has the shorter estimated doubling time?
- Which part of the curve looks most like log-phase growth?
- Why should the chosen time window be reported?

## Common Mistake

- Do not estimate growth rate from the whole curve without considering lag and stationary phases.

## Limitations
