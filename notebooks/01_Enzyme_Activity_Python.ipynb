# 01 Enzyme Activity Summary - Python

## Biochemistry question

In this synthetic enzyme assay, how does enzyme activity compare between the control group and the treatment group?


```python
import plotly.io as pio
pio.renderers.default = "iframe"
```

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv('../data/enzyme_activity_sample.csv')
df
```

```python
summary = (df.groupby('group')
    .agg(mean_activity=('enzyme_activity','mean'),
         sd_activity=('enzyme_activity','std'),
         n=('enzyme_activity','count'))
    .reset_index())
summary['sem_activity'] = summary['sd_activity'] / (summary['n'] ** 0.5)
summary
```

```python
fig = px.bar(summary, x='group', y='mean_activity', error_y='sem_activity',
             title='Enzyme Activity by Group', labels={'mean_activity':'Mean Enzyme Activity'})
# If this chart does not render in Jupyter, try: fig.show(renderer="browser")
fig.show(renderer="iframe")
```

## What you should notice

- The treatment group has a lower mean enzyme activity than the control group in this synthetic assay.
- The error bars summarize uncertainty around each group mean.

## Interpretation practice

- Which group has lower mean enzyme activity?
- What do the error bars help you compare?
- How would you describe the pattern in one cautious sentence?

## Common mistake

- Do not say the treatment "proves" an enzyme mechanism. This plot shows a pattern in a small synthetic dataset.

## Limitations

- This is synthetic learning data, not a real experiment.
- More replicates and experimental context would be needed before making biological claims.
