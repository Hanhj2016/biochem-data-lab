# 18 qPCR Delta-Delta Ct - R

## Biochemistry question

How can synthetic Ct values be normalized to a housekeeping gene and compared as relative expression?

This notebook uses synthetic learning data. Do not treat the output as clinical, diagnostic, regulatory, drug-efficacy, or real biological evidence.


```r
library(tidyverse)

source("../R/qpcr.R")

df <- read_csv("../data/qpcr/qpcr_delta_ct_sample.csv")
df
```

```r
# The helper expects a Ct column name. This dataset uses ct_value.
ct_summary <- mean_ct_by_group(df, ct_col = "ct_value")
ct_summary
```

```r
qpcr_results <- calculate_delta_delta_ct(df, housekeeping_gene = "GAPDH", control_condition = "control", ct_col = "ct_value")
qpcr_results
```

```r
qpcr_results %>%
  filter(condition == "treatment") %>%
  ggplot(aes(x = gene, y = relative_expression)) +
  geom_col(fill = "steelblue") +
  labs(title = "Synthetic qPCR Relative Expression", x = "Gene", y = "Relative expression (2^-ddCt)") +
  theme_minimal()
```

## What You Should Notice

- Lower Ct generally means signal reaches threshold earlier.
- Delta Ct normalizes target genes to a housekeeping gene.
- Delta-delta Ct compares treatment to control.

## Interpretation Practice

- Which treatment gene has the highest relative expression?
- Why is GAPDH used as a housekeeping gene in this teaching example?
- What extra checks would real qPCR need?

## Common Mistake

- Do not interpret raw Ct values without normalization and control checks.

## Limitations

- This is synthetic qPCR data.
- Real qPCR requires primer efficiency, melt curve, no-template control, and replicate QC checks.
