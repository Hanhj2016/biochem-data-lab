## Biochemistry question

How can plate position and edge wells affect interpretation of a synthetic cell viability assay?

This notebook uses synthetic 96-well-style data for learning only.


```python
import plotly.io as pio
pio.renderers.default = "iframe"
```

## 1. Setup and Dataset Preview

The dataset includes row, column, controls, treatment wells, and synthetic cell viability values.


```python
from pathlib import Path
import sys
import pandas as pd
import plotly.express as px

PROJECT_ROOT = Path("..").resolve()
sys.path.append(str(PROJECT_ROOT))

from src.plate_qc import add_plate_position_flags, make_plate_matrix, summarize_edge_effect

df = pd.read_csv(PROJECT_ROOT / "data" / "assay_qc" / "plate_layout_edge_effect.csv")
df.head()
```

## 2. Plate Layout Heatmap

A heatmap helps reveal spatial patterns that a table can hide.


```python
df_qc = add_plate_position_flags(df)
plate_matrix = make_plate_matrix(df_qc)
fig = px.imshow(
    plate_matrix,
    aspect="auto",
    title="Synthetic 96-well Plate QC Heatmap",
    labels={"x": "Column", "y": "Row", "color": "Cell viability (%)"},
)
# If this chart does not render in Jupyter, try: fig.show(renderer="browser")
fig.show(renderer="iframe")
```

## 3. Edge Effect Summary

Compare edge wells with non-edge wells. Edge patterns should be reviewed before interpreting treatment effects.


```python
edge_summary = summarize_edge_effect(df_qc)
edge_summary
```

```python
fig = px.box(
    df_qc,
    x="is_edge_well",
    y="cell_viability_percent",
    color="sample_type",
    points="all",
    title="Edge vs Non-edge Wells",
)
# If this chart does not render in Jupyter, try: fig.show(renderer="browser")
fig.show(renderer="iframe")
```

## Industry / Lab Reality

In plate-based assays, layout can shape the result. Teams review plate maps, edge wells, positive and negative controls, row or column bias, and instrument notes before comparing treatments.

## Career / Job Skill Connection

This notebook practices skills used in assay QC, high-throughput screening, automation support, and biotech/pharma screening workflows: spatial QC, control review, and plate-layout interpretation.


## What You Should Notice

- Plate heatmaps can reveal row, column, or edge patterns.
- Controls help decide whether the plate behaved as expected.
- Edge wells may differ from interior wells in this synthetic example.

## Interpretation Practice

- Which wells are edge wells?
- Do edge wells look different from interior wells?
- What would you check before trusting treatment comparisons?

## Common Mistake

- Do not summarize treatment groups without first checking plate layout and controls.

## Limitations

- This is synthetic learning data, not a real plate-reader result.
- Real QC would need plate maps, protocol notes, instruments, and control acceptance criteria.
