# 15 Enzyme Kinetics: Michaelis-Menten-style Curve - R

## Biochemistry question

How does initial velocity change as substrate concentration increases in a synthetic enzyme kinetics dataset?

This notebook uses synthetic learning data. Do not treat the output as clinical, diagnostic, regulatory, drug-efficacy, or real biological evidence.


```r
library(tidyverse)

source("../R/enzyme_kinetics.R")

clean_df <- read_csv("../data/enzyme_kinetics/michaelis_menten_clean.csv")
noisy_df <- read_csv("../data/enzyme_kinetics/michaelis_menten_noisy.csv")
clean_df
```

```r
clean_summary <- summarize_velocity(clean_df)
clean_summary
```

```r
clean_params <- fit_michaelis_menten(clean_summary)
clean_curve <- make_prediction_table(clean_params, min(clean_summary$substrate_mM), max(clean_summary$substrate_mM))

clean_params[c("vmax", "km")]
```

```r
ggplot(clean_summary, aes(x = substrate_mM, y = mean_velocity)) +
  geom_point(color = "steelblue", size = 2) +
  geom_errorbar(aes(ymin = mean_velocity - sem_velocity, ymax = mean_velocity + sem_velocity), width = 0.05) +
  geom_line(data = clean_curve, aes(y = predicted_velocity), color = "firebrick") +
  labs(title = "Synthetic Michaelis-Menten-style Curve", x = "Substrate (mM)", y = "Mean initial velocity") +
  theme_minimal()
```

```r
noisy_summary <- summarize_velocity(noisy_df) %>% mutate(dataset = "noisy")
clean_plot <- clean_summary %>% mutate(dataset = "clean")
combined <- bind_rows(clean_plot, noisy_summary)

ggplot(combined, aes(x = substrate_mM, y = mean_velocity, color = dataset)) +
  geom_point(size = 2) +
  geom_errorbar(aes(ymin = mean_velocity - sem_velocity, ymax = mean_velocity + sem_velocity), width = 0.05) +
  labs(title = "Clean vs Noisy Synthetic Enzyme Kinetics", x = "Substrate (mM)", y = "Mean initial velocity") +
  theme_minimal()
```

## What You Should Notice

- Velocity rises quickly at low substrate and begins to level off at higher substrate.
- Noisy data can show the same pattern but should make parameter interpretation more cautious.

## Interpretation Practice

- Where does the curve begin to flatten?
- How would you explain `Km` and `Vmax` in plain language?
- Which dataset gives more confidence in the curve shape?

## Common Mistake

- Do not treat fitted `Km` or `Vmax` values from synthetic data as real biochemical constants.

## Limitations

- The fit is educational and simplified.
- Real enzyme kinetics needs experimental controls, initial-rate checks, and model diagnostics.
