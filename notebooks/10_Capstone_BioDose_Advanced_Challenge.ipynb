# 10 Capstone Challenge - BioDose Advanced Extension

## Biochemistry question

How could the synthetic dose-response, QC, and gene-expression notebooks inform a careful educational BioDose AI feature plan?


## Possible Advanced Features

| Feature | Source Notebook | Value |
|---|---|---|
| IC50-style curve fitting | 07 | introduces pharmacology-style modeling |
| Data QC / outlier view | 08 | teaches assay quality thinking |
| Scenario selector | BioDose plan | makes demo more engaging |
| Figure caption generator | BioDose plan | supports academic writing |
| Data quality score | BioDose plan | industry-inspired assay triage |
| Mini poster output | Quarto | portfolio-friendly |

## Design Task

Write a short plan:

1. Which 2–3 advanced features should be added first?
2. What input data do they need?
3. What plot should they generate?
4. What AI explanation should be generated?
5. What warnings or limitations should be shown?
6. How would this help Biochemistry coursework or lab thinking?

## Interpretation Questions

1. Which 2-3 features would make the educational BioDose plan clearer?
2. What input data would each feature need?
3. What plot or table would help a student interpret the result?
4. What warning or limitation should be shown with each feature?

## Limitations

- This capstone is a design exercise using synthetic learning examples.
- A future BioDose demo should not present outputs as clinical, diagnostic, regulatory, or drug-efficacy evidence.
- Any AI-generated explanation should be reviewed against the original data and stated limitations.
