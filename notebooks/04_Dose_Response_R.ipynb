# 04 Dose-Response Visualization - R

## Biochemistry question

In this synthetic cell viability assay, how does measured viability change across compound concentrations?


```r
library(tidyverse)

df <- read_csv('../data/cell_viability_sample.csv')
summary <- df %>% group_by(drug_name, concentration_uM) %>% summarise(mean_viability = mean(cell_viability_percent), sd_viability = sd(cell_viability_percent), n = n(), sem_viability = sd_viability / sqrt(n), .groups='drop')
summary
```

```r
summary %>%
  mutate(plot_concentration = if_else(concentration_uM == 0, 0.001, concentration_uM)) %>%
  ggplot(aes(x = plot_concentration, y = mean_viability, color = drug_name)) +
  geom_line() + geom_point() +
  geom_errorbar(aes(ymin = mean_viability - sem_viability, ymax = mean_viability + sem_viability), width = 0.05) +
  scale_x_log10() +
  labs(title = 'Dose-Response Visualization', x = 'Concentration (uM, log scale; 0 plotted as 0.001)', y = 'Mean Cell Viability (%)')
```

## What you should notice

- Cell viability changes across concentration in this synthetic dose-response example.
- The R plot uses a log-scale x-axis, with zero concentration shown using a small placeholder value.

## Interpretation practice

- Which compound appears to reduce viability more at higher concentrations?
- Which chart details make the dose-response pattern easier to read?
- How would you describe the trend without making a drug-efficacy claim?

## Common mistake

- Do not interpret this plot as evidence that a compound works as a treatment. It is synthetic assay data for learning visualization.

## Limitations

- This is not clinical, diagnostic, regulatory, or drug-efficacy evidence.
- A real dose-response study would need stronger controls, validation, and domain review.
