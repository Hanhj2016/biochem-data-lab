# 03 ANOVA for Multiple Groups - R

## Biochemistry question

In this synthetic cell viability dataset, do any of the group means appear different from the others?


```r
library(tidyverse)

df <- read_csv('../data/anova_cell_viability_sample.csv')
model <- aov(cell_viability ~ group, data = df)
summary(model)
```

```r
summary_df <- df %>% group_by(group) %>% summarise(mean_viability = mean(cell_viability), sd_viability = sd(cell_viability), n = n(), sem_viability = sd_viability / sqrt(n), .groups='drop')

ggplot(summary_df, aes(x = group, y = mean_viability)) +
  geom_col() +
  geom_errorbar(aes(ymin = mean_viability - sem_viability, ymax = mean_viability + sem_viability), width = 0.2) +
  labs(title = 'Cell Viability Across Dose Groups', x = 'Group', y = 'Mean Cell Viability')
```
