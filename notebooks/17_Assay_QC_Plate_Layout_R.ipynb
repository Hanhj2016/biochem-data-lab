# 17 Assay QC Plate Layout - R

## Biochemistry question

Do edge wells in a synthetic plate layout show a spatial pattern that should be reviewed before interpretation?

This notebook uses synthetic learning data. Do not treat the output as clinical, diagnostic, regulatory, drug-efficacy, or real biological evidence.


```r
library(tidyverse)

source("../R/plate_qc.R")

plate_df <- read_csv("../data/assay_qc/plate_layout_edge_effect.csv")
plate_df
```

```r
plate_qc <- add_plate_position_flags(plate_df)
edge_summary <- summarize_edge_effect(plate_qc)
edge_summary
```

```r
ggplot(plate_qc, aes(x = factor(column), y = row, fill = cell_viability_percent)) +
  geom_tile(color = "white") +
  scale_y_discrete(limits = rev(sort(unique(plate_qc$row)))) +
  labs(title = "Synthetic Plate QC Heatmap", x = "Column", y = "Row", fill = "Viability (%)") +
  theme_minimal()
```

```r
ggplot(edge_summary, aes(x = is_edge_well, y = mean_value, fill = sample_type)) +
  geom_col(position = "dodge") +
  labs(title = "Edge vs Interior Summary", x = "Is edge well?", y = "Mean value") +
  theme_minimal()
```

## What You Should Notice

- A plate heatmap can reveal row, column, or edge patterns that a table may hide.
- Edge wells are review points, not automatic exclusions.

## Interpretation Practice

- Do edge and interior wells look different?
- Which sample types are placed near edges?
- What would you check in a real plate map?

## Common Mistake

- Do not interpret a treatment pattern before checking spatial layout.

## Limitations

- This is a synthetic 96-well-style layout.
- Real plate QC needs protocol-specific acceptance criteria and control placement rules.
