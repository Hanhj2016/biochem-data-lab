# 14 Experimental Design and Replicates - R

## Biochemistry question

How do controls, replicate counts, and variability affect interpretation of a synthetic enzyme activity experiment?

This notebook uses synthetic learning data. Do not treat the output as clinical, diagnostic, regulatory, drug-efficacy, or real biological evidence.


```r
library(tidyverse)

source("../R/biochem_stats.R")

df <- read_csv("../data/enzyme_activity/enzyme_activity_three_groups.csv")
df
```

## Dataset Preview

Each row is one synthetic replicate measurement. The `group` column describes the experimental condition.


```r
design_summary <- df %>%
  group_by(group) %>%
  summarise(
    mean_activity = mean(enzyme_activity, na.rm = TRUE),
    sd_activity = sd(enzyme_activity, na.rm = TRUE),
    n = sum(!is.na(enzyme_activity)),
    sem_activity = sd_activity / sqrt(n),
    .groups = "drop"
  )
design_summary
```

```r
ggplot(df, aes(x = group, y = enzyme_activity, color = group)) +
  geom_jitter(width = 0.08, size = 2) +
  stat_summary(fun = mean, geom = "crossbar", width = 0.45, color = "black") +
  labs(title = "Synthetic Enzyme Activity Replicates", x = "Group", y = "Enzyme activity") +
  theme_minimal()
```

```r
design_review <- design_summary %>%
  mutate(
    replicate_note = if_else(n >= 3, "ok for teaching example", "review replicate count"),
    variability_note = if_else(sd_activity > 8, "review variability", "variability modest")
  )
design_review
```

## What You Should Notice

- Individual replicate points help you see spread that a mean alone hides.
- SEM summarizes uncertainty around the mean, while SD describes replicate spread.

## Interpretation Practice

- Which group has the lowest mean activity?
- Are replicate counts balanced across groups?
- What would you check before making a stronger biological interpretation?

## Common Mistake

- Do not treat a lower group mean as proof of enzyme mechanism.

## Limitations

- This is a small synthetic design-review example.
- Real experimental design needs protocol details, randomization, controls, and documentation.
