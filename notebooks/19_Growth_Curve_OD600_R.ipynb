# 19 Growth Curve OD600 - R

## Biochemistry question

How do synthetic OD600 values change over time for two strains, and how can a simple log-phase rate be estimated?

This notebook uses synthetic learning data. Do not treat the output as clinical, diagnostic, regulatory, drug-efficacy, or real biological evidence.


```r
library(tidyverse)

source("../R/growth_curve.R")

df <- read_csv("../data/growth_curve/bacterial_growth_curve.csv")
df
```

```r
growth_summary <- summarize_growth(df)
growth_summary
```

```r
ggplot(growth_summary, aes(x = time_hr, y = mean_od600, color = strain)) +
  geom_line() +
  geom_point() +
  geom_errorbar(aes(ymin = mean_od600 - sem_od600, ymax = mean_od600 + sem_od600), width = 0.2) +
  labs(title = "Synthetic OD600 Growth Curves", x = "Time (hr)", y = "Mean OD600") +
  theme_minimal()
```

```r
rates <- estimate_log_phase_rate(growth_summary, start_hr = 2, end_hr = 6)
rates
```

## What You Should Notice

- OD600 is an indirect measure of culture density.
- Growth-rate estimates depend on the selected time window.

## Interpretation Practice

- Which strain appears to increase faster in the middle of the curve?
- What does doubling time mean?
- Why might a different time window change the estimate?

## Common Mistake

- Do not estimate growth rate from the whole curve without thinking about growth phase.

## Limitations

- This is synthetic OD600 data.
- Real growth curves need blank correction, instrument range checks, and culture metadata.
