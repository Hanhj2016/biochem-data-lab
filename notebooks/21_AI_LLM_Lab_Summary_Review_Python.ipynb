## Industry / Lab Reality

In academic, biotech, QC, or computational life-science settings, AI-written summaries need human review. Useful checks include:

- Is every claim traceable to a dataset, table, or figure?
- Does the summary mention uncertainty, QC, or limitations?
- Does it avoid clinical, diagnostic, regulatory, or drug-efficacy claims?
- Does it distinguish observation from mechanism?
- Does it keep synthetic-data boundaries clear?
