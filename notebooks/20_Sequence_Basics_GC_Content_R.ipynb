# 20 Sequence Basics: GC Content - R

## Biochemistry question

How can short synthetic DNA sequences be summarized by length, GC content, motif count, and reverse complement?

This notebook uses synthetic learning data. Do not treat the output as clinical, diagnostic, regulatory, drug-efficacy, or real biological evidence.


```r
library(tidyverse)

source("../R/sequence_basics.R")

df <- read_csv("../data/sequences/synthetic_sequences.csv")
df
```

```r
seq_summary <- summarize_sequences(df, sequence_col = "dna_sequence") %>%
  mutate(
    atg_count = map_int(dna_sequence, count_motif, motif = "ATG"),
    reverse_complement = map_chr(dna_sequence, reverse_complement)
  )
seq_summary
```

```r
ggplot(seq_summary, aes(x = sequence_id, y = gc_content_percent)) +
  geom_col(fill = "steelblue") +
  labs(title = "Synthetic Sequence GC Content", x = "Sequence ID", y = "GC content (%)") +
  theme_minimal()
```

## What You Should Notice

- DNA sequences can be treated as strings for beginner summaries.
- GC content and motif counts are useful first checks, but they do not prove function.

## Interpretation Practice

- Which sequence has the highest GC content?
- Which sequences contain the `ATG` motif?
- What does reverse complement mean?

## Common Mistake

- Do not infer biological function from one short motif or GC content alone.

## Limitations

- These are short synthetic sequences.
- Real sequence analysis needs quality scores, annotation context, and biological validation.
