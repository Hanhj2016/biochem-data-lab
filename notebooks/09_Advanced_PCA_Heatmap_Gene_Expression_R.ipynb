# 09 Advanced Challenge - PCA and Heatmap for Toy Gene Expression in R

## Biochemistry question

In this toy gene expression matrix, do control, low-treatment, and high-treatment samples show visible expression patterns?


```r
library(tidyverse)

df <- read_csv("../data/gene_expression_matrix_toy.csv")
df
```

```r
expr <- df
rownames(expr) <- expr$gene
expr <- expr %>% select(-gene)
expr_z <- t(scale(t(as.matrix(expr))))

heatmap_df <- as.data.frame(expr_z) %>%
  mutate(gene = rownames(expr_z)) %>%
  pivot_longer(-gene, names_to = "sample", values_to = "z_score")

ggplot(heatmap_df, aes(x = sample, y = gene, fill = z_score)) +
  geom_tile() +
  labs(title = "Toy Gene Expression Heatmap (gene-wise z-score)") +
  theme(axis.text.x = element_text(angle = 45, hjust = 1))
```

```r
# PCA with base R prcomp.
pca <- prcomp(t(expr), scale. = TRUE)

pca_df <- as.data.frame(pca$x[, 1:2]) %>%
  mutate(sample = rownames(pca$x)) %>%
  mutate(group = str_extract(sample, "control|treat_low|treat_high"))

ggplot(pca_df, aes(x = PC1, y = PC2, color = group, label = sample)) +
  geom_point(size = 3) +
  geom_text(vjust = -0.8) +
  labs(title = "Toy PCA Sample Map")
```
