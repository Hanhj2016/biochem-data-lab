# 18 qPCR Delta-Delta Ct - Python

## Biochemistry question

How can Ct values be normalized to a housekeeping gene to compare relative expression in synthetic qPCR data?

This notebook uses synthetic qPCR data for learning only. It is not diagnostic or clinical data.


```python
import plotly.io as pio
pio.renderers.default = "notebook_connected"
```

## 1. Setup and Dataset Preview


```python
from pathlib import Path
import sys
import pandas as pd
import plotly.express as px

PROJECT_ROOT = Path("..").resolve()
sys.path.append(str(PROJECT_ROOT))
from src.qpcr import calculate_delta_delta_ct, mean_ct_by_group

df = pd.read_csv(PROJECT_ROOT / "data" / "qpcr" / "qpcr_delta_ct_sample.csv")
df.head()
```

## 2. Mean Ct Values

Lower Ct means the fluorescence threshold was reached earlier. Interpretation should happen after normalization.


```python
ct_summary = mean_ct_by_group(df)
ct_summary
```

## 3. Delta-Delta Ct and Relative Expression


```python
results = calculate_delta_delta_ct(df, housekeeping_gene="GAPDH", control_condition="control")
results.round(3)
```

```python
plot_df = results[results["condition"] == "treatment"]
fig = px.bar(plot_df, x="gene", y="relative_expression", title="Synthetic qPCR Relative Expression", labels={"relative_expression":"Relative expression (2^-ddCt)"})
# If this chart does not render in Jupyter, try: fig.show(renderer="iframe") or fig.show(renderer="browser")
fig.show()
```

## What You Should Notice

- Ct values are normalized to the housekeeping gene before comparing targets.
- Treatment samples show higher relative expression for some genes and lower for others.
- The housekeeping gene should be stable for this teaching calculation to make sense.

## Interpretation Practice

- Which target gene has the highest treatment relative expression?
- Why do we subtract the housekeeping Ct?
- What could go wrong if the housekeeping gene is not stable?

## Common Mistake

- Do not compare raw Ct values alone as if they were normalized expression.

## Limitations

- This is synthetic learning data, not diagnostic or clinical qPCR data.
- Real qPCR analysis requires primer validation, efficiency checks, controls, and replicate review.
