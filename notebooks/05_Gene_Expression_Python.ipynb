# 05 Toy Gene Expression Fold Change - Python

## Biochemistry question

In this toy gene expression dataset, which genes show higher or lower expression in the treatment samples compared with controls?


```python
import plotly.io as pio
pio.renderers.default = "iframe"
```

```python
import pandas as pd
import numpy as np
import plotly.express as px

df = pd.read_csv('../data/gene_expression_toy.csv')
df.head()
```

```python
control_cols = ['control_1','control_2','control_3']
treatment_cols = ['treatment_1','treatment_2','treatment_3']

df['control_mean'] = df[control_cols].mean(axis=1)
df['treatment_mean'] = df[treatment_cols].mean(axis=1)
df['fold_change'] = df['treatment_mean'] / df['control_mean']
df['log2_fold_change'] = np.log2(df['fold_change'])
summary = df[['gene','control_mean','treatment_mean','fold_change','log2_fold_change']]
summary.sort_values('log2_fold_change', ascending=False)
```

```python
fig = px.bar(summary.sort_values('log2_fold_change'), x='gene', y='log2_fold_change', title='Toy Gene Expression: log2 Fold Change')
# If this chart does not render in Jupyter, try: fig.show(renderer="browser")
fig.show(renderer="iframe")
```

## What you should notice

- Log2 fold change shows which genes are higher or lower in treatment samples compared with controls.
- The bar chart helps compare genes quickly, but it is still a toy example.

## Interpretation practice

- Which genes are higher in the treatment samples?
- Which genes are lower in the treatment samples?
- How would you explain log2 fold change to another student?

## Common mistake

- Do not treat fold change alone as proof that a gene has a biological role. It is one descriptive summary.

## Limitations

- This is a toy synthetic gene expression dataset, not a full RNA-seq workflow.
- A real workflow would need normalization choices, quality control, statistics, and validation.
