## 3. QC Analyst Track: Plate Batch Review

Career/lab question: do plate batches show control-range problems or spatial patterns that need review before interpretation?


````python
qc = pd.read_csv(DATA_DIR / "qc_plate_batch_review.csv")
qc_summary = summarize_qc_plate_batches(qc)
qc_summary.head(12)
````

````python
fig = px.bar(
    qc_summary,
    x="sample_type",
    y="mean_viability",
    color="review_status",
    facet_col="batch_id",
    error_y="sem_viability",
    title="QC Analyst Track: Synthetic Plate Batch Summary",
    labels={"mean_viability": "Mean cell viability (%)"},
)
# If this chart does not render in Jupyter, try: fig.show(renderer="browser")
fig.show(renderer="iframe")
````

## Academic Advancement Practice

Choose one track and prepare an application-ready artifact:

```text
one figure or table + one cautious interpretation + one limitation + one next-step idea + one short reflection
```

Interview prompt: which review flag mattered most, and how would you explain it to a research mentor, program interviewer, or lab supervisor?

Good boundary sentence:

```text
This project used synthetic learning data to practice lab-data reasoning and should not be interpreted as clinical, diagnostic, regulatory, drug-efficacy, or real biological evidence.
```


## What You Should Notice

- Career and academic readiness often comes from careful review, not only from complex models.
- QC flags, missing replicates, day shifts, Ct spread, ambiguous bases, and AI overclaims are all examples of judgment points.
- A strong student artifact should explain what was checked, what was observed, what is limited, and what should happen next.

## Interpretation Practice

- Which track best matches your academic or career goal?
- Which dataset has the clearest quality concern?
- Which figure or table would be easiest to explain in an interview?

## Common Mistake

- Do not present a role-style synthetic exercise as real lab experience or real biological evidence. Present it as practice in reproducible analysis, quality review, and scientific communication.

## Limitations

- These are small synthetic scenarios, not full industry workflows.
- Real labs use project-specific SOPs, acceptance criteria, instruments, metadata, and supervisor review.
