# 16 Standard Curve: Bradford-style Protein Assay - R

## Biochemistry question

How can a standard curve estimate unknown protein concentrations from synthetic absorbance values?

This notebook uses synthetic learning data. Do not treat the output as clinical, diagnostic, regulatory, drug-efficacy, or real biological evidence.


```r
library(tidyverse)

source("../R/standard_curve.R")

df <- read_csv("../data/standard_curves/bradford_standard_curve.csv")
df
```

```r
standard_summary <- summarize_standard_curve(df)
standard_summary
```

```r
fit <- fit_linear_standard_curve(df)
fit[c("slope", "intercept", "r_squared", "min_standard", "max_standard")]
```

```r
ggplot(standard_summary, aes(x = known_concentration_mg_ml, y = mean_absorbance)) +
  geom_point(color = "steelblue", size = 2) +
  geom_errorbar(aes(ymin = mean_absorbance - sem_absorbance, ymax = mean_absorbance + sem_absorbance), width = 0.03) +
  geom_abline(intercept = fit$intercept, slope = fit$slope, color = "firebrick") +
  labs(title = "Synthetic Bradford-style Standard Curve", x = "Known concentration (mg/mL)", y = "Mean absorbance at 595 nm") +
  theme_minimal()
```

```r
unknown_estimates <- estimate_unknown_concentrations(df, fit)
unknown_summary <- summarize_unknown_estimates(unknown_estimates)
unknown_summary
```

## What You Should Notice

- Standards with known concentrations define the calibration line.
- Unknowns outside the standard range should be reviewed before reporting.

## Interpretation Practice

- Which unknown has an estimate outside the standard range?
- Why is interpolation safer than extrapolation?
- What would a real lab do before trusting an above-range unknown?

## Common Mistake

- Do not report an extrapolated estimate as if it were equally reliable.

## Limitations

- This is a synthetic calibration example.
- Real Bradford assays require blank correction, dilution details, and protocol-specific checks.
