# 05 Toy Gene Expression Fold Change - R

## Biochemistry question

In this toy gene expression dataset, which genes show higher or lower expression in the treatment samples compared with controls?


```r
library(tidyverse)

df <- read_csv('../data/gene_expression_toy.csv')
df
```

```r
summary <- df %>%
  rowwise() %>%
  mutate(control_mean = mean(c_across(control_1:control_3)),
         treatment_mean = mean(c_across(treatment_1:treatment_3)),
         fold_change = treatment_mean / control_mean,
         log2_fold_change = log2(fold_change)) %>%
  ungroup() %>%
  select(gene, control_mean, treatment_mean, fold_change, log2_fold_change)
summary %>% arrange(desc(log2_fold_change))
```

```r
summary %>% arrange(log2_fold_change) %>%
  ggplot(aes(x = gene, y = log2_fold_change)) +
  geom_col() +
  labs(title = 'Toy Gene Expression: log2 Fold Change', x = 'Gene', y = 'log2 Fold Change') +
  theme(axis.text.x = element_text(angle = 45, hjust = 1))
```

## What you should notice

- Log2 fold change shows which genes are higher or lower in treatment samples compared with controls.
- The R workflow calculates row-wise control and treatment means before plotting.

## Interpretation practice

- Which genes are higher in the treatment samples?
- Which genes are lower in the treatment samples?
- Which part of the R workflow makes the fold-change calculation easiest to follow?

## Common mistake

- Do not treat fold change alone as proof that a gene has a biological role. It is one descriptive summary.

## Limitations

- This is a toy synthetic gene expression dataset, not a full RNA-seq workflow.
- A real workflow would need normalization choices, quality control, statistics, and validation.
